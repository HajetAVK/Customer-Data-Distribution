```python
import pandas as pd
df = pd.read_csv('data_test.csv')
display(df.head())
```

```python
print("Информация о данных")
df.info()
```

```python
print("Количество пропусков")
print(df.isnull().sum())
```

```python
print("Статистика")
print(df.describe())
df_range = df['variable'].max() - df['variable'].min()
print(f'range    {df_range}')
```

```python
import matplotlib.pyplot as plt
from scipy import stats

fig, axes = plt.subplots(1, 2, figsize=(12,5))

axes[0].hist(df, bins=8, edgecolor='black', alpha=0.7)
axes[0].set_title('Распределение')
axes[0].set_xlabel('Число')
axes[0].set_ylabel('Частота')

axes[1].boxplot(df)
axes[1].set_title('Ящик с усами')
axes[1].set_ylabel('Число')

plt.tight_layout()
plt.show()

# Проверка на нормальность
stat, p_value = stats.shapiro(df['variable'])
#.4f  Означает 4 знака после запятой, тип float
print(f"Статистика теста: {stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value > 0.05:
    print("Вывод: p-value > 0.05. Данные распределены нормально.")
else:
    print("Вывод: p-value < 0.05. Данные НЕ распределены нормально.")
```

В данных существует выброс в районе 60. Он делает распределение данных ненормальным, что влияет на последующее тестирование. 
Основная масса значений лежит в промежутке от ~5 до ~40.
Данные имеют правостороннюю ассиметрию из-за наличия выброса.
